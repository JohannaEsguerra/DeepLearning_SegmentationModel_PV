```python
import os
import cv2
import numpy as np
from PIL import Image
import torch
import torchvision.transforms as transforms
from UNet import UNet  
```

```python
# Define dataset directories
test_data_dir = r'C:\Users\32esla1mpg\Documents\UNet_PV\test_data'

# Create a transform to preprocess the images
transform = transforms.Compose([
        transforms.ToTensor()
])

# Initialize U-Net model
model = UNet(in_channels=3, out_channels=1)  # Modify based on your model architecture
model.load_state_dict(torch.load(r'C:\Users\32esla1mpg\Documents\UNet_PV\outputs\trained_PV.pth'))  # Load the trained model
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)
model.eval()

# Create a list to store predicted masks
predicted_masks = []
```

```python
# Iterate over the images in the test dataset
for img_name in os.listdir(test_data_dir):
    img_path = os.path.join(test_data_dir, img_name)
    image = Image.open(img_path).convert('RGB')
    image_tensor = transform(image).unsqueeze(0).to(device)
    
    with torch.no_grad():
        output = model(image_tensor)
            
        # Debugging: Print the range of the output before sigmoid
        print(f"Output range before sigmoid: min {output.min().item()}, max {output.max().item()}")
            
        predicted_mask = torch.sigmoid(output).squeeze(0).cpu().numpy()
            
        # Debugging: Print the range of the predicted mask after sigmoid
        print(f"Predicted mask range after sigmoid: min {predicted_mask.min()}, max {predicted_mask.max()}")
            
        # Thresholding to create binary mask
        binary_mask = (predicted_mask > 0.5).astype(np.uint8)
            
        # Append the binary mask to the list
        predicted_masks.append(binary_mask)
       
# Convert the list of predicted masks to a numpy array
predicted_masks_array = np.array(predicted_masks)

# Create a directory to save the predicted masks
output_dir = r'C:\Users\32esla1mpg\Documents\UNet_PV\outputs\masks_predicted'
os.makedirs(output_dir, exist_ok=True)
```

```python
 # Iterate over the predicted masks and save each mask as an image file
for i, mask in enumerate(predicted_masks_array):
    # Construct the file name for the mask (e.g., mask_1.png, mask_2.png, etc.)
    mask_file = os.path.join(output_dir, f'mask_{i+1}.png')
    
    # Ensure the mask is binary (0 and 1) and save as an image file using OpenCV
    if len(mask.shape) == 3:
        mask = mask.squeeze()
        
    # Save the binary mask directly (values 0 and 1)
    cv2.imwrite(mask_file, mask * 255)  # Multiply by 255 for visualization as 0-255 image

# To verify, print the shape of a sample predicted mask
print(f'Sample predicted mask shape: {predicted_masks_array[0].shape}')
```
