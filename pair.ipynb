This code help you to check if the model is seeing the image with the corresponding mask.

```python
import torch
from torch.utils.data import DataLoader
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
from torch.utils.data import Dataset, DataLoader # creating custom datasets and loading data in batches.
# Import the SolarPanelSegmentationDataset class from the dataloader file
from dataloader import SolarPanelSegmentationDataset
from PIL import Image 
import os 


# Visualization function
def visualize_sample(image, mask, title):
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(image.permute(1, 2, 0))  # Convert from (C, H, W) to (H, W, C)
    axes[0].set_title('Image')
    axes[1].imshow(mask.squeeze(), cmap='gray')  # Assuming mask is (1, H, W)
    axes[1].set_title('Mask')
    plt.suptitle(title)
    plt.show()

# Debugging function
def debug_sample(dataset):
    for i in range(len(dataset)):
        image, mask, filename = dataset[i]
        image_path = dataset.image_filenames[i]
        mask_path = dataset.mask_filenames[i]

        print(f"Sample {i} - Filename: {filename}")
        print(f"Image path: {image_path}, Mask path: {mask_path}")

        visualize_sample(image, mask, f"Sample {i} - Filename: {filename}")

        if i == 5:  # Print only the first 5 samples
            break

# Define transformation for preprocessing
transform = transforms.Compose([
    transforms.ToTensor()
])

# Define paths to the image and mask directories
image_dir = r'C:\Users\32esla1mpg\Documents\UNet_PV\train_data\images'
mask_dir = r'C:\Users\32esla1mpg\Documents\UNet_PV\train_data\masks'

# Create the dataset
dataset = SolarPanelSegmentationDataset(image_dir, mask_dir, transform=transform)

# Create a dataloader
batch_size = 4
dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

# Run the debug function
debug_sample(dataset)

# Check correspondence by filenames
image_filenames = sorted(os.listdir(image_dir))
mask_filenames = sorted(os.listdir(mask_dir))

for img_filename, mask_filename in zip(image_filenames, mask_filenames):
    print(f"Image: {img_filename}, Mask: {mask_filename}")
    if img_filename != mask_filename:
        print("Mismatch found!")

# Visualize some pairs from the dataloader
image_batch, mask_batch, filenames_batch = next(iter(dataloader))
for i in range(min(len(image_batch), 4)):  # Visualize up to 4 samples
    visualize_sample(image_batch[i], mask_batch[i], filenames_batch[i])
```
